## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/hassan9039/intern1/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Total pages:", len(df))
print("Mean CTR:", round(df["ctr"].mean(), 3))
print("Mean impressions (90d):", round(df["impressions_90d"].mean(), 1))
```

The starter dataset contains 30,000 pages, giving enough observations to investigate page-level prioritization. The mean CTR is 0.511, while pages receive an average of about 5,200 impressions over 90 days. These numbers suggest that the dataset contains substantial search activity and measurable page-performance signals that could be useful for prioritizing pages for human review. This does not yet show that a refresh will improve performance, but it gives a reasonable basis for investigating which pages may deserve attention.
